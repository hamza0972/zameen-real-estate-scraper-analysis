### extract city

```python
# define a function to extract the city
def city(x):
  city=x.split(',')[1]
  return city
```

### extract location

```python
# define a function to extract the location
def location(x):
  location=x.split(',')[0]
  return location
```

### remove PKR from price

```python
def price_value(x):
  price=x.split('PKR')[1]
  return price
```

### Apply these functions

```python
df['city']=df['location'].apply(city)
# on running it, it will show error as there are some rows that donot have ',' in location
# check them
```

```python
# check the rows that donot have ',' in the location
df[df['location'].apply(lambda x: ',' not in str(x))]['location']
```

```python
# remove them
df = df[df['location'] != 'For Sale']
df = df.reset_index(drop=True)
```

```python
# now apply
df['city']=df['location'].apply(city)
```

```python
df['location']=df['location'].apply(location)
```

```python
df['price']=df['price'].apply(price_value)
```

```python
df
```

### now convert lakh, crore and arab in numeric price

```python
df['price'].apply(lambda x: 'Crore' in str(x)).sum()
```

```python
# check the rows that donot contain price in Crore
df[df['price'].apply(lambda x: 'Crore' not in str(x))]
```

```python
# create a function to convert
def price_int(x):
  price=x.split(' ')
  if price[1] == 'Lakh':
    price_numeric=float(price[0]) * 100000
  elif price[1] == 'Crore':
    price_numeric=float(price[0]) * 10000000
  else:
    price_numeric=float(price[0]) * 1000000000
  return price_numeric
```

```python
df['price']=df['price'].apply(price_int)
```

```python
df
```

### convert Kanal to Marla

```python
# first check that all is all the data related to kanal and marla
first=df['Area'].apply(lambda x: 'Kanal' not in str(x)).sum()
second=df['Area'].apply(lambda x: 'Marla' not in str(x)).sum()
print(first+second)
```

```python
# define a function to convert kanal to marla
def Kanal_to_marla(x):
  value_unit=x.split(' ')
  value=value_unit[0]
  unit=value_unit[1]
  if unit == 'Kanal':
    new_value = float(value) * 20
  else:
    new_value = float(value)
  return new_value
```

```python
df['Area']=df['Area'].apply(Kanal_to_marla)
```
